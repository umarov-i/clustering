```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
data = pd.read_csv('3.01.+Country+clusters.csv')
data
```

```python
plt.scatter(data['Longitude'], data['Latitude'])
plt.xlim(-180, 180)
plt.ylim(-90, 90)
plt.show()
```

```python
x = data.iloc[:, 1:3]
x
```

```python
kmeans = KMeans(2)
```

```python
kmeans.fit(x)
```

```python
identified_clusters = kmeans.fit_predict(x)
identified_clusters
```

```python
data_clusters = data.copy()
data_clusters['Clusters'] = identified_clusters
data_clusters
```

```python
plt.scatter(data['Longitude'], data['Latitude'], c=data_clusters['Clusters'], cmap='rainbow')
plt.xlim(-180, 180)
plt.ylim(-90, 90)
plt.show()
```

```python
data_mapped = data.copy()
data_mapped['Language'] = data_mapped['Language'].map({'English': 0, 'French': 1, 'German': 2})
data_mapped
```

```python
x = data_mapped.iloc[:, 3:4]
x
```

```python
kmeans_lang = KMeans(3)
kmeans.fit(x)
```

```python
cluster_lang = kmeans_lang.fit_predict(x)
cluster_lang
```

```python
data_cluster_lang = data_mapped.copy()
data_cluster_lang['Cluster'] = cluster_lang
data_cluster_lang
```

```python
plt.scatter(data_cluster_lang['Longitude'], data_cluster_lang['Latitude'], c=data_cluster_lang['Cluster'], cmap='rainbow')
plt.xlim(-180, 180)
plt.ylim(-90, 90)
plt.show()
```
