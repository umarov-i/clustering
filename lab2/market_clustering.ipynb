```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
data = pd.read_csv('3.12.+Example.csv')
data
```

```python
plt.scatter(data['Satisfaction'], data['Loyalty'])
plt.xlabel('Satisfaction')
plt.ylabel('Loyalty')
plt.show()
```

```python
x = data.copy()
```

```python
kmeans = KMeans(2)
kmeans.fit(x)
```

```python
clusters = x.copy()
clusters['cluster'] = kmeans.fit_predict(x)
```

```python
plt.scatter(clusters['Satisfaction'], clusters['Loyalty'], c=clusters['cluster'], cmap='rainbow')
plt.xlabel('Satisfaction')
plt.ylabel('Loyalty')
plt.show()
```

```python
from sklearn import preprocessing
x_scaled = preprocessing.scale(x)
x_scaled
```

```python
wcss = []

for i in range(1, 10):
    kmeans = KMeans(i)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(1, 10), wcss)
plt.xlabel('number of clusters')
plt.ylabel('wcss')
plt.show()
```

```python
kmeans2 = KMeans(4)
kmeans2.fit(x_scaled)
clusters2 = x.copy()
clusters2['cluster'] = kmeans2.fit_predict(x_scaled)
clusters2
```

```python
plt.scatter(clusters2['Satisfaction'], clusters2['Loyalty'], c=clusters2['cluster'], cmap='rainbow')
plt.xlabel('Satisfaction')
plt.ylabel('Loyalty')
plt.show()
```
